# Credit Risk Default Prediction

The goal of this project is to predict loan default risk using borrower and loan characteristics.

## Problem Statement

The objective of this project is to predict whether a loan will default based on borrower and loan characteristics.
This is a binary classification problem commonly encountered in banking and financial risk management.

## Project Workflow

1. Load and inspect the data
2. Clean and preprocess features
3. Perform exploratory data analysis (EDA)
4. Train baseline and machine learning models
5. Evaluate performance using appropriate metrics
